```python
import pandas as pd
data = {'item':['pen','book','bag'],'price':[5,20,50],'qty':[10,3,2]}
df = pd.DataFrame(data)
df
```

```python
df['price']
df['price']*df['qty']
df['revenue'] = df['price']*df['qty']
df
```

```python
df['double_price'] = df['price']*2
df
```

```python
df[df['price'] > 10]
```

```python
df[df['price'] == 5]
df[df['qty']>=3]
df[df['item']=='pen']
```

```python
df[df['qty']<5]
df[df['item']!='bag']
```

```python
df[df['item'].str.startswith('b')]
df[-df['item'].str.startswith('b')]
```

```python
df[df['item'].str.startswith('p')]
df[-df['item'].str.startswith('p')]
```

```python
import pandas as pd
sales = pd.DataFrame({'country': ['UK', 'France', 'UK', 'France', 'Spain'], 'revenue': [100, 50, 200, 30, 80]})
sales
```

```python
sales.groupby('country')['revenue'].sum()
```

```python
sales.groupby('country')['revenue'].sum().sort_values(ascending=False)
```

```python
sales.groupby('country')['revenue'].mean()
```

```python
sales.groupby('country')['revenue'].count()
```

```python
dates = pd.DataFrame({'date': pd.to_datetime(['2024-01-15', '2024-01-20', '2024-02-03']), 'revenue': [100, 50, 80]})
dates['month'] = dates['date'].dt.month
dates
```

```python
dates.groupby('month')['revenue'].sum()
```

```python
import matplotlib.pyplot as plt
months = [1, 2, 3]
revenue = [100, 150, 120]
plt.bar(months, revenue)
plt.title('Revenue by month')
plt.xlabel('Month')
plt.ylabel('Revenue (GBP)')
plt.show()
```
