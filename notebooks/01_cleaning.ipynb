```python
df_clean = df.copy()
df_clean.shape
```

```python
df_clean.drop_duplicates(inplace=True)
```

```python
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])
```

```python
df_clean.info()
```

```python
returns = df_clean[df_clean['Invoice'].str.startswith('C')]
returns.shape
```

```python
df_clean = df_clean[~df_clean['Invoice'].str.startswith('C')]
df_clean.shape
```

```python
df_clean = df_clean[df_clean['Price'] > 0]
df_clean.shape
```

```python
df_clean['Revenue'] = df_clean['Price'] * df_clean['Quantity']
df_clean.head()
```

```python
df_clean = df_clean[ ~ df_clean['Quantity'].isin([ 80995, 74215 ]) ]
df_clean.shape
```

```python
non_product_codes = ['POST', 'DOT', 'M', 'm', 'C2', 'D', 'S', 'B']
non_products = df_clean[df_clean['StockCode'].isin(non_product_codes)]
df_clean = df_clean[~df_clean['StockCode'].isin(non_product_codes)]
print(non_products.shape, df_clean.shape)
```

```python
df_clean[df_clean['StockCode'].str.len() >= 8][['StockCode', 'Description']].drop_duplicates()
```

```python
extra = (df_clean['StockCode'].isin(['BANK CHARGES', 'AMAZONFEE']) | df_clean['StockCode'].str.startswith('gift_', na=False)).astype(bool)
non_products = pd.concat([non_products, df_clean[extra]])
df_clean = df_clean[~extra]
print(non_products.shape, df_clean.shape)
```

```python
df_clean.to_csv('../data/clean_sales.csv', index=False)
```
